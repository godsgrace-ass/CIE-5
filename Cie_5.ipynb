```python
import spacy
from nltk import ngrams
nlp = spacy.load("en_core_web_sm")
doc = nlp("Ai is a simulation of human intelligance by machine.It includes expects system NLP speech recognition amd machine vision")
tokens = [token.text for token in doc]
bigrams = list(ngrams(tokens, 2))
print(bigrams)
```
